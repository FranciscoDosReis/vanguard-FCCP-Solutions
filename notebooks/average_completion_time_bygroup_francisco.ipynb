## **Clients completed with errors**

```python
#Completed not perfect condition function
required_steps = {"start", "step_1", "step_2", "step_3", "confirm"}

def has_repeated_step_complete(steps):
    counts = steps.value_counts()
    

    has_all_steps = all(counts.get(step, 0) >= 1 for step in required_steps)
    
    has_repetition = any(counts.get(step, 0) > 1 for step in required_steps)
    
    return has_all_steps and has_repetition
```

```python
mask_test = df_test.groupby("client_id")["process_step"].apply(has_repeated_step_complete)
client_ids_test = mask_test[mask_test].index
df_repeat_complete_test = df_test[df_test["client_id"].isin(client_ids_test)].copy()


df_repeat_complete_test["time_dt"] = pd.to_datetime(df_repeat_complete_test["time"], format="%H:%M:%S")
client_duration_repeat_test = (
    df_repeat_complete_test.groupby("client_id")["time_dt"]
    .agg(min_time="min", max_time="max")
    .reset_index()
)
client_duration_repeat_test["time_spent"] = (
    client_duration_repeat_test["max_time"] - client_duration_repeat_test["min_time"]
)

avg_sec_test = client_duration_repeat_test["time_spent"].dt.total_seconds().mean()
med_sec_test = client_duration_repeat_test["time_spent"].dt.total_seconds().median()

print("--- TEST GROUP (Complete with Repetitions) ---")
print(f"Total clients analyzed: {len(client_duration_repeat_test)}")
print(f"Average time spent: {pd.to_timedelta(avg_sec_test, unit='s')}")
print(f"Median time spent: {pd.to_timedelta(med_sec_test, unit='s')}\n")
```

```python
required_steps = ["start", "step_1", "step_2", "step_3", "confirm"]

def get_segments(df):

    counts = (
        df.groupby(["client_id", "process_step"]).size()
        .unstack(fill_value=0)
        .reindex(columns=required_steps, fill_value=0)
    )
    has_all = (counts >= 1).all(axis=1)
    has_repeat = (counts > 1).any(axis=1)

    return {
        "Full completion":       counts.index[has_all],
        "Perfect run":           counts.index[has_all & ~has_repeat],
        "Completed with errors": counts.index[has_all & has_repeat],
        "Not completed":         counts.index[~has_all],
    }

segments_test = get_segments(df_test)
segments_control = get_segments(df_control)


for name, seg in [("Test", segments_test), ("Control", segments_control)]:
    print(f"--- {name} ---")
    for label, ids in seg.items():
        print(f"{label}: {len(ids)} clients")
    assert len(seg["Perfect run"]) + len(seg["Completed with errors"]) == len(seg["Full completion"])
    print()
```

## Each step time spent - overall (not effective)

```python
step_order = ["start", "step_1", "step_2", "step_3", "confirm"]


gaps_test = build_gaps(df_test)
gaps_control = build_gaps(df_control)
segments_test = get_segments(df_test)
segments_control = get_segments(df_control)

def step_durations(gaps_df):
    d = gaps_df.sort_values(["client_id", "timestamp"], kind="stable").copy()

    d["next_ts"] = d.groupby("client_id")["timestamp"].shift(-1)
    d["step_time"] = d["next_ts"] - d["timestamp"]
    return d.dropna(subset=["step_time"])[["client_id", "process_step", "step_time"]]
```

```python
from IPython.display import display

seg_order = list(segments_test.keys())

def summarize_steps(gaps_df, segments, group_name, per="visit"):
    sd = step_durations(gaps_df)
    out = []
    for seg_name, ids in segments.items():
        s = sd[sd["client_id"].isin(ids)]
        if per == "client":
            # add up repeated visits to the same step, one value per client and step
            s = s.groupby(["client_id", "process_step"], as_index=False)["step_time"].sum()
        stats = (
            s.groupby("process_step")["step_time"]
            .agg(["count", "mean", "median"])
            .reset_index()
        )
        stats.insert(0, "segment", seg_name)
        stats.insert(0, "group", group_name)
        out.append(stats)
    return pd.concat(out, ignore_index=True)

def build_step_table(per):
    t = pd.concat([
        summarize_steps(gaps_test, segments_test, "Test", per),
        summarize_steps(gaps_control, segments_control, "Control", per),
    ], ignore_index=True)

    t["process_step"] = pd.Categorical(t["process_step"], categories=step_order, ordered=True)
    t["segment"] = pd.Categorical(t["segment"], categories=seg_order, ordered=True)
    t[["mean", "median"]] = t[["mean", "median"]].apply(lambda c: c.dt.round("s"))

    return (
        t.rename(columns={"count": f"n_{per}s", "mean": "mean_time", "median": "median_time"})
         .sort_values(["segment", "process_step", "group"])
         .reset_index(drop=True)
    )

step_summary_visit = build_step_table("visit")
step_summary_client = build_step_table("client")

print("PER VISIT")
display(step_summary_visit)
print("PER CLIENT")
display(step_summary_client)
```
