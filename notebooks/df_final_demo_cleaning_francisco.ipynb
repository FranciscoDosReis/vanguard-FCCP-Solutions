```python
import pandas as pd
```

```python
df_final_demo = pd.read_csv("../data/raw/df_final_demo.csv")
```

```python
df_final_demo.info()
```

```python
df_final_demo.columns = df_final_demo.columns.str.strip().str.lower()
```

```python
tenure = df_final_demo[["clnt_tenure_yr", "clnt_tenure_mnth"]]
tenure
```

```python
unique_yr = df_final_demo["clnt_tenure_yr"].unique()
unique_yr
```

```python
unique_mnth = df_final_demo["clnt_tenure_mnth"].unique()
unique_mnth
```

```python
df_final_demo['clnt_tenure_yr'] = df_final_demo['clnt_tenure_yr'].round().astype("Int64")
df_final_demo['clnt_tenure_mnth'] = df_final_demo['clnt_tenure_mnth'].round().astype("Int64")
```

```python
df_final_demo
```

```python
unique_num_acc = df_final_demo["num_accts"].unique()
unique_num_acc
```

```python
df_final_demo['num_accts'] = df_final_demo['num_accts'].round().astype("Int64")
```

```python
df_final_demo
```

```python
unique_calls_6_mnth = df_final_demo["calls_6_mnth"].unique()
unique_calls_6_mnth
```

```python
df_final_demo.info()
```

```python
unique_gender = df_final_demo["gendr"].unique()
unique_gender
```

```python
gender_counts = df_final_demo["gendr"].value_counts(dropna=False)

print(gender_counts)
```

```python
#X and Nan not statistically relevant - change to U - Unspecified"
```

```python
df_final_demo["gendr"] = df_final_demo["gendr"].fillna("U").replace("X", "U")
```

```python
gender_counts = df_final_demo["gendr"].value_counts(dropna=False)

print(gender_counts)
```

```python
df_final_demo["gendr"] = df_final_demo["gendr"].replace("M", "Male")
df_final_demo["gendr"] = df_final_demo["gendr"].replace("F", "Female")
df_final_demo["gendr"] = df_final_demo["gendr"].replace("U", "Unspecified")
```

```python
gender_counts = df_final_demo["gendr"].value_counts(dropna=False)

print(gender_counts)
```

```python
df_final_demo = df_final_demo.rename(columns={
    "gendr": "gender",
    "clnt_tenure_yr": "client_tenure_yr",
    "clnt_tenure_mnth": "client_tenure_mnth",
    "clnt_age": "client_age",
    "bal": "total_balance",
})
```

```python
df_final_demo
```

```python
unique_age = df_final_demo["client_age"].unique()
unique_age
```

```python
df_final_demo['client_age'] = df_final_demo['client_age'].round().astype("Int64")

df_final_demo
```

```python
calls_6_mnth_un = df_final_demo["calls_6_mnth"].unique()
calls_6_mnth_un
```

```python
df_final_demo['calls_6_mnth'] = df_final_demo['calls_6_mnth'].round().astype("Int64")
df_final_demo['logons_6_mnth'] = df_final_demo['logons_6_mnth'].round().astype("Int64")

df_final_demo
```
