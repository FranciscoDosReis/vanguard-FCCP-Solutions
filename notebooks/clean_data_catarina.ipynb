```python
import pandas as pd

df = pd.read_csv("../data/raw/df_final_web_data_pt_1.csv")
df.head(50)
```

client_id: The Vanguard client identification number

visitor_id: The device/browser they used. One client can have several if they used a laptop and a phone.

visit_id: The session they are in. Each time they come to the site and start browsing, a new visit_id is created.

process_step: Which step they reached: start, step_1, step_2, step_3 or confirm

date_time: The timestamp when the process_step happenned

Rows are events, not steps reached, because the same client can hit the same step several times in one session,

Each row is one event at a specific timestamp (date_time), a specific client (client_id), using a particular device/browser (visitor_id), in a particular session (visit_id), reached a specific step of the process (process_step).
